```python
import json, csv
import re
import pandas as pd
from sklearn.metrics import r2_score
```

### Graph PageRank

```python

json_file_path = '/home/ubuntu/memFigLessDIR/Parrotfish-UBC/results-graph-pagerank.json'
# json_file_path = '/home/ubuntu/memFigLessDIR/Mem-Fig-Less/estimation-algorithms/results-graph-pagerank-v1.5.json'
with open(json_file_path, 'r') as file:
    json_data = json.load(file)

parsed_data = (json_data)
parsed_data = [x for x in parsed_data if 'processing_time' in x]
# Parse the JSON data
parsed_data = sorted(parsed_data, key=lambda x: x['payload']['n'])
# parsed_data = [x for x in parsed_data if 'init_duration' in x and x['init_duration'] is None]
payload = []
memory = []

for item in parsed_data:
    payload.append(item['payload']['n'])
    # Define the regular expression pattern
    pattern = r"Optimization result: (\d+)\.\d+ MB"

    # Search for the pattern in the string
    match = re.search(pattern, item['result'])

    # Extract the integer if a match is found
    if match:
        result = int(match.group(1))
        # print(f'Extracted integer: {result}')
    else:
        print('No match found for ipnut:', item['payload']['n'])
        result = 0
    memory.append(result)


import matplotlib.pyplot as plt

plt.figure(figsize=(6, 6))
plt.plot(payload, memory, label='Optimised Memory Configuration')
# plt.plot(payload, [1305 for _ in payload], label='Execution Time Limit')
plt.ylabel('Memory (MB)', fontsize=16, fontweight='bold')
plt.xlabel('Payload (n)', fontsize=16, fontweight='bold')
# plt.title('Time vs Payload', fontsize=16, fontweight='bold')
plt.legend(fontsize=14)
plt.show()
```
